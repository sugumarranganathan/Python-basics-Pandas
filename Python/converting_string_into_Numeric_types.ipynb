**convert a string into an integer or float using int() and float().**

```python
str_num = '56'

print(int(str_num))       # default base is 10
print(int(str_num, 16))   # From Hexadecimal
print(int(str_num, 12))   # From a number where base is 12

str_num = '25.897'

print(float(str_num))     # convert string to floating point value
```

    int(str_num, 16)
    print(int(str_num, 16))

    Here '56' is being interpreted as a hexadecimal number.

    Hexadecimal:

    56₁₆

    Calculation:

    5 × 16 + 6 × 1
    = 80 + 6
    = 86

    So:

    int('56', 16)

    produces:

    86

    ====

    Important

    The 16 does not mean "multiply by 16."

    It means:

    Interpret the string as a base-16 number.

```python
| Code                 | Explanation                                                                         |
| -------------------- | ----------------------------------------------------------------------------------- |
| `str_num = '56'`     | Stores `56` as a **string** because it is inside quotes.                            |
| `int(str_num)`       | Converts the string `'56'` into an integer using **base 10**.                       |
| `int(str_num, 16)`   | Interprets `'56'` as a **base-16 (hexadecimal)** number and converts it to decimal. |
| `int(str_num, 12)`   | Interprets `'56'` as a **base-12** number and converts it to decimal.               |
| `str_num = '25.897'` | Stores a decimal number as a string.                                                |
| `float(str_num)`     | Converts the string `'25.897'` into a floating-point number.                        |
```
