for loop — Definition

A for loop takes items one by one from a collection and performs an action for each item.

```python
numbers = [2, 4, 6, 8]

for i in numbers:
    print(i)
```

How it works
for i in numbers:

means:

"Take each item from numbers one by one and temporarily store it in i."

---

| Iteration | `i` |
| --------- | --: |
| 1st       |   2 |
| 2nd       |   4 |
| 3rd       |   6 |
| 4th       |   8 |

---

    Another example
    for i in [2, 4, 6]:
        print(i * 2)

    Output:

    4
    8
    12

    So:

    for → repeat for each item
    i → current item
    in → take items from
    list → collection of items

    ----

    Loop = the overall repetition process
    Iteration = one repetition of that loop

```python
List = [2, 4, 6, 7, 8]

[M * 2 for M in List]
```

    single process, multiple inputs (for loop)

    The loop takes each value:

    M = 2 → 2 × 2 = 4
    M = 4 → 4 × 2 = 8
    M = 6 → 6 × 2 = 12
    M = 7 → 7 × 2 = 14
    M = 8 → 8 × 2 = 16

| Part            | Meaning                               |
| --------------- | ------------------------------------- |
| `[ ]`           | Create a **new list**                 |
| `M * 2`         | Multiply each item by 2               |
| `for M in List` | Take each item from `List` one by one |
