**This example shows how to convert an ASCII number into its corresponding character using the Python chr() function.**

```python
a = chr(65)
b = chr(77)

print(a)
print(b)
```

    What is chr()?

    chr() converts an integer ASCII/Unicode code point into its corresponding character.

    For example:

    chr(65)

    → A

    chr(77)

    → M

    Output
    A
    M

ASCII values

| Number | Character |
| -----: | :-------: |
|   `65` |    `A`    |
|   `66` |    `B`    |
|   `67` |    `C`    |
|   `68` |    `D`    |
|   `69` |    `E`    |
|    ... |    ...    |
|   `77` |    `M`    |
|   `90` |    `Z`    |


| Code          | Explanation                                                        |
| ------------- | ------------------------------------------------------------------ |
| `a = chr(65)` | Converts ASCII value `65` into its character and stores it in `a`. |
| `b = chr(77)` | Converts ASCII value `77` into its character and stores it in `b`. |
| `print(a)`    | Prints the character stored in `a`.                                |
| `print(b)`    | Prints the character stored in `b`.                                |
