Converting Containers

This example shows how to convert between List, Set, and Tuple using Python's list(), set(), and tuple() functions.

```python
my_list = [10, 20, 30, 40, 50]
my_set = {10, 20, 30, 20, 50, 20}
my_tuple = (10, 25, 50, 75, 100,10)

print('list to tuple: ' + str(tuple(my_list)))
print('list to set: ' + str(set(my_list)))
print('set to list: ' + str(list(my_set)))
print('set to tuple: ' + str(tuple(my_set)))
print('tuple to list: ' + str(list(my_tuple)))
print('tuple to set: ' + str(set(my_tuple)))
```

All conversions together

| Conversion   | Code             | Result type |
| ------------ | ---------------- | ----------- |
| List → Tuple | `tuple(my_list)` | Tuple       |
| List → Set   | `set(my_list)`   | Set         |
| Set → List   | `list(my_set)`   | List        |
| Set → Tuple  | `tuple(my_set)`  | Tuple       |
| Tuple → List | `list(my_tuple)` | List        |
| Tuple → Set  | `set(my_tuple)`  | Set         |


One important thing to remember: converting List/Tuple → Set can remove duplicate values and the order is not guaranteed.
