String to Character Converting and Base Converting

This example teaches two things:

Character → ASCII value using ord()
Decimal number → different bases using hex(), oct(), and bin()

```python
print('ASCII value of "M" is: ' + str(ord('M')))
print('Hexadecimal value of 254 is: ' + str(hex(254)))
print('Octal value of 62 is: ' + str(oct(62)))
print('Binary value of 56 is: ' + str(bin(56)))
```

| Code       | Function | Meaning                          | Output     |
| ---------- | -------- | -------------------------------- | ---------- |
| `ord('M')` | `ord()`  | Character → ASCII/Unicode number | `77`       |
| `hex(254)` | `hex()`  | Decimal → Hexadecimal (base 16)  | `0xfe`     |
| `oct(62)`  | `oct()`  | Decimal → Octal (base 8)         | `0o76`     |
| `bin(56)`  | `bin()`  | Decimal → Binary (base 2)        | `0b111000` |

----

| Part | Meaning                    |
| ---- | -------------------------- |
| `0x` | Indicates hexadecimal      |
| `fe` | Hexadecimal representation |


| Function   | Conversion            |
| ---------- | --------------------- |
| `ord('M')` | Character → Number    |
| `chr(77)`  | Number → Character    |
| `hex(254)` | Decimal → Hexadecimal |
| `oct(62)`  | Decimal → Octal       |
| `bin(56)`  | Decimal → Binary      |
| `str(77)`  | Number → String       |
