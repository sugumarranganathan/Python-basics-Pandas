**This example shows Set → Dictionary conversion using zip() and dict().**

```python
list_keys = {1, 2, 3, 4}
list_values = {'Mon', 'Tue', 'Wed', 'Thu'}

new_dict = dict(zip(list_keys, list_values))

print(new_dict)
```

| Code                                         | Explanation                                                              |
| -------------------------------------------- | ------------------------------------------------------------------------ |
| `list_keys = {1, 2, 3, 4}`                   | Creates a **Set** of numbers. These will become the dictionary **keys**. |
| `list_values = {'Mon', 'Tue', 'Wed', 'Thu'}` | Creates a **Set** of days. These will become the dictionary **values**.  |
| `zip(list_keys, list_values)`                | Pairs one key with one value.                                            |
| `dict(...)`                                  | Converts those pairs into a **Dictionary**.                              |
| `new_dict = ...`                             | Stores the new dictionary.                                               |
| `print(new_dict)`                            | Prints the dictionary.                                                   |


```python
list_keys = [1, 2, 3, 4]
list_values = ['Mon', 'Tue', 'Wed', 'Thu']

new_dict = dict(zip(list_keys, list_values))

print(new_dict)
```

**This is another method to convert a Set into a Dictionary, using dict.fromkeys().**

```python
list_keys = {1, 2, 3, 4}

new_dict = dict.fromkeys(list_keys, 'Mon')

print(new_dict)
```

| Code                       | Meaning                                                 |
| -------------------------- | ------------------------------------------------------- |
| `list_keys = {1, 2, 3, 4}` | Creates a **Set** containing `1, 2, 3, 4`.              |
| `dict.fromkeys(...)`       | Creates a dictionary using the given items as **keys**. |
| `list_keys`                | Becomes the dictionary's **keys**.                      |
| `'Mon'`                    | Becomes the **same value for every key**.               |
| `print(new_dict)`          | Prints the dictionary.                                  |

===

    Easy formula to remember
    dict.fromkeys(keys, value)

    keys → dictionary keys
    value → same value for all keys
