      What is "Implicit"?

      Implicit = automatic.

      You did not tell Python to convert 12 to a float.

      You simply wrote:

      new_num = num1 + num2

      Python automatically handles the conversion.

      Compare

| Type of conversion      | Meaning                       | Example             |
| ----------------------- | ----------------------------- | ------------------- |
| **Implicit conversion** | Python converts automatically | `12 + 1.23 → 13.23` |
| **Explicit conversion** | You tell Python to convert    | `float(12) → 12.0`  |


```python
num1 = 12
num2 = 1.23

new_num = num1 + num2

# display new value and resulting data type
print("Value:", new_num)
print("DataType:", type(new_num))
```

**Addition of string and integer using Explicit conversion**

```python
num_str = '12'
num_int = 23

# explicit type conversion
num_str = int(num_str)

num_sum = num_int + num_str

print("Sum:", num_sum)
print("DataType:", type(num_sum))
```

| Code                          | What it means                                         |
| ----------------------------- | ----------------------------------------------------- |
| `num_str = '12'`              | `'12'` is a **string**, because it is inside quotes.  |
| `num_int = 23`                | `23` is an **integer (`int`)**.                       |
| `num_str = int(num_str)`      | Explicitly converts `'12'` from **string → integer**. |
| `num_sum = num_int + num_str` | Adds `23 + 12`.                                       |
| `print("Sum:", num_sum)`      | Prints the sum.                                       |
| `type(num_sum)`               | Checks the data type of the result.                   |


    Why do we need explicit conversion?

    If you tried:

    num_str = '12'
    num_int = 23

    num_sum = num_int + num_str

    Python would give an error because:

    integer + string

    cannot be directly added.

    You need to explicitly convert:

    int('12')

    Then:

    23 + 12 = 35

Implicit vs Explicit
--------------------

| Implicit Conversion                              | Explicit Conversion                             |
| ------------------------------------------------ | ----------------------------------------------- |
| Python converts automatically                    | You tell Python to convert                      |
| No conversion function needed                    | Uses functions like `int()`, `float()`, `str()` |
| Example: `12 + 1.23`                             | Example: `int('12')`                            |
| `int` → `float` automatically during calculation | `str` → `int` manually                          |
