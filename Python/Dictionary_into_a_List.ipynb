    This example shows how to convert a Dictionary into a List containing
    
    key-value pairs using list() and .items().

    ----------

    Take all key-value pairs from the dictionary and convert them into a list.

```python
inputDictionary = {'Hello': 10, 'HopeAI': 20, 'python': 30}

resultList = list(inputDictionary.items())

print(resultList)
```

| Code                      | Explanation                                      |
| ------------------------- | ------------------------------------------------ |
| `inputDictionary = {...}` | Creates a **dictionary** with key-value pairs.   |
| `inputDictionary.items()` | Gets **both keys and values together** as pairs. |
| `list(...)`               | Converts those pairs into a **list**.            |
| `resultList = ...`        | Stores the resulting list.                       |
| `print(resultList)`       | Displays the list.                               |


convert a Dictionary into a List containing

key-value pairs Using **keys()** Method

```python
inputDictionary = {'Hello': 10, 'HopeAI': 20, 'python': 30}

resultList = list(inputDictionary.keys())

print(resultList)
```

convert a Dictionary into a List containing

key-value pairs Using **values ()** Method

```python
inputDictionary = {'Hello': 10, 'HopeAI': 20, 'python': 30}

resultList = list(inputDictionary.values())

print(resultList)
```

**dictionary into a list using List Comprehension.**

```python
inputDictionary = {'Hello': 10, 'HopeAI': 20, 'python': 30}

resultList = [(key, value) for key, value in inputDictionary.items()]

print(resultList)
```

**This is the zip() method for converting a dictionary into a list of key-value tuples.**

```python
inputDictionary = {'Hello': 10, 'HopeAI': 20, 'python': 30}

resultList = zip(inputDictionary.keys(), inputDictionary.values())

resultList = list(resultList)

print(resultList)
```

**This example shows how to use the map() function to convert each dictionary key-value pair into a list.**

```python
inputDictionary = {'Hello': 10, 'HopeAI': 20, 'python': 30}

resultList = new_list = list(map(list, inputDictionary.items()))

print(resultList)
```
