**This example shows how to convert a Tuple into a List using the list() function.**

```python
Course = ('Machine Learning', 'Python', 'Data Science', 'Deep Learning')

Course_Names = list(Course)

print(Course_Names)
```

| Code                  | Explanation                                  |
| --------------------- | -------------------------------------------- |
| `Course = (...)`      | Creates a **Tuple** containing course names. |
| `( )`                 | Parentheses represent a Tuple.               |
| `list(Course)`        | Converts the Tuple into a **List**.          |
| `Course_Names = ...`  | Stores the converted List.                   |
| `print(Course_Names)` | Prints the List.                             |
