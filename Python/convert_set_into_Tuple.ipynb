```python
Course = {'Machine Learning', 'Python', 'Data Science', 'Deep Learning'}

Course_Names = tuple(Course)

print(Course_Names)
```

    Set → unordered
    Tuple → ordered
    tuple(set) → converts the set to a tuple, but does not automatically sort it.

| Code                  | Explanation                                                                           |
| --------------------- | ------------------------------------------------------------------------------------- |
| `Course = {...}`      | Creates a **Set** containing course names.                                            |
| `{ }`                 | Curly braces are used for a set when there are only values and no `key: value` pairs. |
| `tuple(Course)`       | Converts the set into a **tuple**.                                                    |
| `Course_Names = ...`  | Stores the new tuple.                                                                 |
| `print(Course_Names)` | Prints the tuple.                                                                     |

---

Set vs Tuple

| Feature    | Set                   | Tuple              |
| ---------- | --------------------- | ------------------ |
| Symbol     | `{ }`                 | `( )`              |
| Example    | `{'Python', 'AI'}`    | `('Python', 'AI')` |
| Ordered?   | ❌ No guaranteed order | ✅ Ordered          |
| Duplicates | ❌ Not allowed         | ✅ Allowed          |
| Mutable?   | ✅ Yes                 | ❌ No               |

```python
Course = {'Machine Learning', 'Python', 'Data Science', 'Deep Learning'}

Course_Names = tuple(sorted(Course))

print(Course_Names)
```

**This example is showing Set → List Comprehension → Tuple.**

```python
Course = {'Machine Learning', 'Python', 'Data Science', 'Deep Learning'}

Course_Names = [i for i in Course]

print(tuple(Course_Names))
```
