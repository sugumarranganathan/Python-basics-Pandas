Primitive datatype conversion means changing a basic data type from one type to another.

===

| Word           | Meaning                                               |
| -------------- | ----------------------------------------------------- |
| **Primitive**  | Basic/simple data types                               |
| **Datatype**   | The type of data, such as `int`, `float`, `str`, etc. |
| **Conversion** | Changing one datatype into another datatype           |


converting the NumPy array values from float to integer using astype(int).

```python
import numpy as np

x = np.array([1, 2, 2.5])

x.astype(int)
```

| Code                        | Meaning                                                     |
| --------------------------- | ----------------------------------------------------------- |
| `import numpy as np`        | Imports the NumPy library and gives it the short name `np`. |
| `x = np.array([1, 2, 2.5])` | Creates a NumPy array and stores it in `x`.                 |
| `x.astype(int)`             | Converts the array's values to **integer (`int`)** type.    |

====

Why does 2.5 become 2?

When converting float → int, the decimal/fractional part is removed.

| Original | After `astype(int)` |
| -------: | ------------------: |
|    `1.0` |                 `1` |
|    `2.0` |                 `2` |
|    `2.5` |                 `2` |


**This is the Integer → Float conversion example.**

```python
# int to float
x, y = 3, 4

z = float(x + y)

print(z)
```

      x = 3       → int
      y = 4       → int
          ↓
      x + y
          ↓
      7           → int
          ↓
      float(7)
          ↓
      7.0         → float

**Float → Integer**

```python
# float to int
x, y = 3.5, 4.8

z = int(x + y)

print(z)
```

| Code              | Meaning                              |
| ----------------- | ------------------------------------ |
| `x, y = 3.5, 4.8` | `x` and `y` are **float** values.    |
| `x + y`           | Adds them: `3.5 + 4.8 = 8.3`         |
| `int(x + y)`      | Converts `8.3` from **float → int**. |
| `z = ...`         | Stores the result in `z`.            |
| `print(z)`        | Prints `8`.                          |

==

    3.5 + 4.8
        ↓
      8.3       ← float
        ↓
    int(8.3)
        ↓
      8        ← int

**Integer → String**

```python
# int to str
chocolate = 25
icecream = 40

total = chocolate + icecream

print("The total is: " + str(total) + "$")
```

| Code                           | Meaning                                   |
| ------------------------------ | ----------------------------------------- |
| `chocolate = 25`               | `25` is an **integer**.                   |
| `icecream = 40`                | `40` is an **integer**.                   |
| `total = chocolate + icecream` | Calculates `25 + 40 = 65`.                |
| `str(total)`                   | Converts integer `65` into string `"65"`. |
| `+ "$"`                        | Adds the dollar symbol to the string.     |
| `print()`                      | Displays the final message.               |


| Conversion       | Code    | Example    | Result |
| ---------------- | ------- | ---------- | ------ |
| **Float → Int**  | `int()` | `int(8.3)` | `8`    |
| **Int → String** | `str()` | `str(65)`  | `"65"` |
