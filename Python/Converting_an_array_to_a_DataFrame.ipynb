converting an array to a DataFrame usually means taking data stored in a NumPy array and putting it into a Pandas DataFrame so that it becomes easier to analyze, filter, clean, and visualize.

1. Simple 1D array

```python
import numpy as np
import pandas as pd

arr = np.array([10, 20, 30, 40, 50])

df = pd.DataFrame(arr, columns=["Marks"])

print(df)
```

Here:

    arr → NumPy array
    pd.DataFrame() → converts it into a DataFrame
    columns=["Marks"] → gives the column name
    0, 1, 2... → Pandas automatically creates the index

2. 2D array

```python
arr = np.array([
    [25, 50000],
    [30, 60000],
    [35, 70000]
])

df = pd.DataFrame(arr, columns=["Age", "Salary"])

print(df)
```

converting a 2D NumPy array into a Pandas DataFrame.

```python
import numpy as np
import pandas as pd

marks = np.array([[11,22,33],[44,55,66]])

df = pd.DataFrame(
    marks,
    columns=['Subject1', 'Subject2', 'Subject3']
)

print(df)
```

    NumPy Array
        ↓
    [[11,22,33],
    [44,55,66]]
        ↓
    pd.DataFrame()
        ↓
    Pandas DataFrame
        ↓
    Subject1  Subject2  Subject3
      11        22        33
      44        55        66

Key syntax to remember:

df = pd.DataFrame(array, columns=['Column1', 'Column2', 'Column3'])

This example is the same Array → DataFrame conversion, but now it also assigns custom row/index names.

The important new part is:

index=['Anu', 'John']

```python
import numpy as np
import pandas as pd

marks = np.array([[11,22,33],[44,55,66]])

df = pd.DataFrame(
    marks,
    columns=['Subject1', 'Subject2', 'Subject3'],
    index=['Anu', 'John']
)

print(df)
```

    What does index mean?

    Normally, Pandas automatically gives row numbers:

      Subject1  Subject2  Subject3
    0         11        22        33
    1         44        55        66

    But here:

    index=['Anu', 'John']

    you are telling Pandas:

    "Instead of 0 and 1, use Anu and John as the row labels."

    So:

    Array                         DataFrame

    [11, 22, 33]       →          Anu   →  11  22  33
    [44, 55, 66]       →          John  →  44  55  66

    Remember these 3 parameters
    pd.DataFrame(
        marks,
        columns=['Subject1', 'Subject2', 'Subject3'],
        index=['Anu', 'John']
    )

    | Parameter | Purpose      |
    | --------- | ------------ |
    | `marks`   | Data/array   |
    | `columns` | Column names |
    | `index`   | Row names    |


    Simple memory trick:

    columns → names for columns
    index → names for rows

This Example 2 is a little more advanced because the array contains different

types of data: names, ages, birth years, and graduation years.

```python
import numpy as np
import pandas as pd

student_details = np.array([
    ['Jon', 25, 1995, 2016],
    ['Maria', 47, 1973, 2000],
    ['Bill', 38, 1982, 2005]
], dtype=object)

df = pd.DataFrame(
    student_details,
    columns=['Name', 'Age', 'Birth Year', 'Graduation Year']
)

print(df)
```

    1. Creating the NumPy array
    student_details = np.array([
        ['Jon', 25, 1995, 2016],
        ['Maria', 47, 1973, 2000],
        ['Bill', 38, 1982, 2005]
    ], dtype=object)

    The data looks like a table:

    Name     Age    Birth Year    Graduation Year
    Jon      25        1995             2016
    Maria    47        1973             2000
    Bill     38        1982             2005

    There are:

    3 rows → 3 students
    4 columns → Name, Age, Birth Year, Graduation Year
    2. Why dtype=object?

    This is important.

    The array contains both:

    'Jon'      → string
    25         → integer
    1995       → integer
    2016       → integer

    So the array contains different data types.

    By specifying:

    dtype=object

    we tell NumPy:

    "Allow each element to keep its own Python data type."

    3. Convert Array → DataFrame
    df = pd.DataFrame(
        student_details,
        columns=['Name', 'Age', 'Birth Year', 'Graduation Year']
    )

    Here:

    student_details
          ↓
      NumPy array
          ↓
    pd.DataFrame()
          ↓
      DataFrame

    The columns parameter gives names to the four columns.

    4. Output
        Name  Age  Birth Year  Graduation Year
    0    Jon   25        1995             2016
    1  Maria   47        1973             2000
    2   Bill   38        1982             2005

    Notice that Pandas automatically creates:

    0
    1
    2

    as the row/index values because we didn't provide a custom index.

    5. Why is this useful?

    Once the data is a DataFrame, you can easily perform operations such as:

    df['Name']

    Get only the names.

    df['Age']

    Get the ages.

    df[df['Age'] > 30]

    Get students whose age is greater than 30.

    df.shape

    Output:

    (3, 4)

    Meaning 3 rows × 4 columns.

    One easy trick to remember

    Quotes → String

    "Jon"
    "Chennai"
    "25"

    No quotes + whole number → Integer

    25
    100
    1995

    And with decimal:

    25.5

    that is a Float, not an Integer.

This example is teaching you how **to check the DataFrame type and the data type of each column**

```python
import numpy as np
import pandas as pd

student_details = np.array([
    ['Jon', 25, 1995, 2016],
    ['Maria', 47, 1973, 2000],
    ['Bill', 38, 1982, 2005]
], dtype=object)

df = pd.DataFrame(
    student_details,
    columns=['Name', 'Age', 'Birth Year', 'Graduation Year']
)

print(df)
print(type(df))
print(df.dtypes)
```

    Name               object
    Age                object
    Birth Year         object
    Graduation Year    object
    dtype: object

    Why object?

    Because NumPy stored the mixed data ("Jon", 25, 1995, etc.) as an object-type array.

The better Pandas approach

If you want the columns to have their proper types, create the DataFrame and then convert the numeric columns:

```python
df['Age'] = df['Age'].astype(int)
df['Birth Year'] = df['Birth Year'].astype(int)
df['Graduation Year'] = df['Graduation Year'].astype(int)
```

```python
print(df.dtypes)
```

    Now:

    Name → object (text)
    Age → int64 (integer)
    Birth Year → int64 (integer)
    Graduation Year → int64 (integer)

    ----

    Age, Birth Year, and Graduation Year are integer columns, using 64-bit representation.

    ----

    int32 vs int64

    Both are integers:

    int32 → integer stored using 32 bits
    int64 → integer stored using 64 bits

    ----

    What does 64 mean?

    The 64 means 64 bits of computer memory are used to store each integer value.

    A bit is a tiny unit of computer information that can be either:

    0 or 1

    So:

    64 bits = 64 binary positions

| Data type | Meaning                            |
| --------- | ---------------------------------- |
| `int64`   | Integer, 64-bit                    |
| `float64` | Decimal number, 64-bit             |
| `object`  | General Python objects, often text |
| `bool`    | True / False                       |


This example is teaching a very important Pandas concept:

How to add a NumPy array as a new column to an existing DataFrame.

```python
import numpy as np
import pandas as pd

#create pandas DataFrame
df = pd.DataFrame({
    'Game1': [25, 12, 15, 14, 19, 23, 25, 29],
    'Game2': [5, 7, 7, 9, 12, 9, 9, 4],
    'Game3': [11, 8, 10, 6, 6, 5, 9, 12]
})

#create NumPy array for 'Game4'
Game4 = np.array([2, 3, 1, 0, 2, 7, 8, 2])

#add 'Game4' array as new column in DataFrame
df['Game4'] = Game4.tolist()

#display the DataFrame
print(df)
```

| Code                                         | Explanation                                                        |
| -------------------------------------------- | ------------------------------------------------------------------ |
| `import numpy as np`                         | Imports the **NumPy** library and gives it the short name `np`.    |
| `import pandas as pd`                        | Imports the **Pandas** library and gives it the short name `pd`.   |
| `#create pandas DataFrame`                   | Comment explaining that we are going to create a Pandas DataFrame. |
| `df = pd.DataFrame({...})`                   | Creates a **Pandas DataFrame** and stores it in the variable `df`. |
| `'Game1': [25, 12, 15, 14, 19, 23, 25, 29]`  | Creates the **Game1 column** with 8 values.                        |
| `'Game2': [5, 7, 7, 9, 12, 9, 9, 4]`         | Creates the **Game2 column** with 8 values.                        |
| `'Game3': [11, 8, 10, 6, 6, 5, 9, 12]`       | Creates the **Game3 column** with 8 values.                        |
| `Game4 = np.array([2, 3, 1, 0, 2, 7, 8, 2])` | Creates a **NumPy array** called `Game4` containing 8 values.      |
| `df['Game4']`                                | Refers to a column named **Game4** in the DataFrame.               |
| `=`                                          | Assigns a value to the `Game4` column.                             |
| `Game4.tolist()`                             | Converts the NumPy array into a **Python list**.                   |
| `df['Game4'] = Game4.tolist()`               | Adds the `Game4` values as a **new column** in the DataFrame.      |
| `print(df)`                                  | Displays the complete DataFrame.                                   |
