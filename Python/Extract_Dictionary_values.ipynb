    This example shows how to extract only the values from a dictionary and convert them into a list.


```python
my_dict = {'A': 300, 'B': 150, 'C': 500, 'D': 850, 'E': 400}

my_list = list(my_dict.values())

print(my_list)
```

| Code               | Explanation                                          |
| ------------------ | ---------------------------------------------------- |
| `my_dict = {...}`  | Creates a **dictionary** containing key-value pairs. |
| `my_dict.values()` | Gets **only the values** from the dictionary.        |
| `list(...)`        | Converts those values into a **list**.               |
| `my_list = ...`    | Stores the new list in `my_list`.                    |
| `print(my_list)`   | Prints the list.                                     |
