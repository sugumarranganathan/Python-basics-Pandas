This example shows how to use the ord() function to convert a single character

into its ASCII/Unicode numeric value.



```python
a = '+'
b = 'A'

print(ord(a))
print(ord(b))
```

| Code      | Explanation                                 |
| --------- | ------------------------------------------- |
| `a = '+'` | Stores the `+` character as a **string**.   |
| `b = 'A'` | Stores the `A` character as a **string**.   |
| `ord(a)`  | Converts `+` into its ASCII/Unicode number. |
| `ord(b)`  | Converts `A` into its ASCII/Unicode number. |
| `print()` | Displays the numeric values.                |


===

| Character |    `ord()` | Numeric value |
| --------- | ---------: | ------------: |
| `'+'`     | `ord('+')` |          `43` |
| `'A'`     | `ord('A')` |          `65` |
