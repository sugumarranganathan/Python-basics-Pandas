Convert to Complex Number

This example shows how to create a complex number in Python using the complex() function.

```python
my_complex = complex(10, 5)

print(my_complex)
```

| Code         | Meaning                                         |
| ------------ | ----------------------------------------------- |
| `complex()`  | Python function used to create a complex number |
| `10`         | **Real part**                                   |
| `5`          | **Imaginary part**                              |
| `my_complex` | Variable storing the complex number             |
| `print()`    | Displays the result                             |


How complex(10, 5) works

Python represents a complex number as:

a + bj

where:

| Part | Meaning        |
| ---- | -------------- |
| `a`  | Real part      |
| `b`  | Imaginary part |
| `j`  | Imaginary unit |


| Code             | Output    |
| ---------------- | --------- |
| `complex(10, 5)` | `(10+5j)` |
| `complex(3, 2)`  | `(3+2j)`  |
| `complex(7, 0)`  | `(7+0j)`  |
| `complex(0, 4)`  | `4j`      |


      10 → Real part
      5 → Imaginary part
      j → Imaginary unit
      ↓
      10 + 5j
