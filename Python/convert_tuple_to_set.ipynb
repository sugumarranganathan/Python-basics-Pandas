**This example shows how to convert a Tuple into a Set using the set() function.**

```python
Course = ('Machine Learning', 'Python', 'Data Science', 'Deep Learning')

Course_Names = set(Course)

print(Course_Names)
```

| Code                  | Explanation                                  |
| --------------------- | -------------------------------------------- |
| `Course = (...)`      | Creates a **Tuple** containing course names. |
| `( )`                 | Parentheses represent a tuple.               |
| `set(Course)`         | Converts the tuple into a **Set**.           |
| `Course_Names = ...`  | Stores the new Set.                          |
| `print(Course_Names)` | Prints the Set.                              |
