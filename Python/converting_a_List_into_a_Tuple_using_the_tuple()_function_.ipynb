converting a List into a Tuple using the tuple() function.

```python
Marks = [45, 98, 23, 76, 38, 62]
Marks_tuple = tuple(Marks)
print(Marks_tuple)
```

| Code                               | Explanation                                               |
| ---------------------------------- | --------------------------------------------------------- |
| `Marks = [45, 98, 23, 76, 38, 62]` | Creates a **list** of marks. `[ ]` represents a list.     |
| `tuple(Marks)`                     | Converts the list `Marks` into a **tuple**.               |
| `Marks_tuple = ...`                | Stores the converted tuple in the variable `Marks_tuple`. |
| `print(Marks_tuple)`               | Displays the tuple.                                       |

===

Take the list → convert it into a tuple.

For your example:

    Marks = [45, 98, 23, 76, 38, 62]

    Marks_tuple = tuple(Marks)
    List
    [45, 98, 23, 76, 38, 62]
              ↓
          tuple()
              ↓
    Tuple
    (45, 98, 23, 76, 38, 62)

This is another way to convert a List into a Tuple

using a for loop inside tuple().

```python
Marks = [45, 98, 23, 76, 38, 62]

Marks_tuple = tuple(i for i in Marks)

print(Marks_tuple)
```

To multiply each mark by 2 and then convert the result into a tuple,

```python
Marks = [45, 98, 23, 76, 38, 62]

Marks_tuple = tuple(i * 2 for i in Marks)

print(Marks_tuple)
```

    i → each mark
    i * 2 → operation
    for i in Marks → takes each mark one by one
    tuple() → converts the results into a tuple.

    ----

    tuple(expression for i in list)
